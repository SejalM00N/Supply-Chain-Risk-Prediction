## Investigating Data Leakage for Classification Target

Given the classification models achieved perfect accuracy, we need to investigate potential data leakage. Features that are highly correlated with, or directly derived from, the target variable can lead to artificially inflated performance. We will examine `disruption_likelihood_score`, `delay_probability`, and `route_risk_level` as potential leaking features for `risk_classification`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert encoded labels back to original for plotting
data['risk_classification_decoded'] = le.inverse_transform(y_clf_encoded)

# Visualize potential leaking features against risk_classification
fig, axes = plt.subplots(1, 3, figsize=(21, 6))

sns.boxplot(x='risk_classification_decoded', y='disruption_likelihood_score', data=data, ax=axes[0])
axes[0].set_title('Risk Classification vs. Disruption Likelihood Score')
axes[0].set_xlabel('Risk Classification')
axes[0].set_ylabel('Disruption Likelihood Score')

sns.boxplot(x='risk_classification_decoded', y='delay_probability', data=data, ax=axes[1])
axes[1].set_title('Risk Classification vs. Delay Probability')
axes[1].set_xlabel('Risk Classification')
axes[1].set_ylabel('Delay Probability')

sns.boxplot(x='risk_classification_decoded', y='route_risk_level', data=data, ax=axes[2])
axes[2].set_title('Risk Classification vs. Route Risk Level')
axes[2].set_xlabel('Risk Classification')
axes[2].set_ylabel('Route Risk Level')

plt.tight_layout()
plt.show()
```

### Conclusion on Data Leakage

The visualizations clearly show a strong and almost perfect separation of the `risk_classification` categories based on `disruption_likelihood_score`, `delay_probability`, and `route_risk_level`. This indicates that these features are highly correlated with, or directly derived from, the target variable, leading to data leakage. To build a robust and generalizable model, these features must be removed from the training data.

## Retraining Classification Models without Leaking Features

```python
# Drop the identified leaking features from the original feature set 'x'
leaking_features = ['disruption_likelihood_score', 'delay_probability', 'route_risk_level']
x_clf_no_leak = x.drop(columns=leaking_features)

# Scale the new feature set
scaler_no_leak = StandardScaler()
x_scaled_no_leak = scaler_no_leak.fit_transform(x_clf_no_leak)

# Re-split the data into training and testing sets
x_train_clf_no_leak, x_test_clf_no_leak, y_train_clf_no_leak, y_test_clf_no_leak = \
    train_test_split(x_scaled_no_leak, y_clf_encoded, test_size=0.2, random_state=42)

# Re-apply SMOTE to the training data without leaking features
smote_no_leak = SMOTE(random_state=42)
x_train_clf_resampled_no_leak, y_train_clf_resampled_no_leak = smote_no_leak.fit_resample(x_train_clf_no_leak, y_train_clf_no_leak)

print("Shape of x_train_clf_resampled_no_leak:", x_train_clf_resampled_no_leak.shape)
print("Shape of y_train_clf_resampled_no_leak:", y_train_clf_resampled_no_leak.shape)

print("\nClass distribution after SMOTE (no leaking features):")
print(pd.Series(y_train_clf_resampled_no_leak).value_counts())
```

```python
print("--- Retraining Classification Models (no leaking features) ---")
for name, model in classification_models.items():
  print(f"\nTraining {name} with resampled data (no leaking features)....")
  model.fit(x_train_clf_resampled_no_leak, y_train_clf_resampled_no_leak)
  y_pred_clf_no_leak = model.predict(x_test_clf_no_leak)
  accuracy_no_leak = accuracy_score(y_test_clf_no_leak, y_pred_clf_no_leak)
  print(f"{name} Accuracy (no leaking features): {accuracy_no_leak: .2f}")
  print(f"Classification Report for {name} (no leaking features):\n{classification_report(y_test_clf_no_leak, y_pred_clf_no_leak, target_names=le.classes_ )}")
```

## Select and Save the Best Classification Model

```python
import joblib

# Identify the best model based on the retraining results
best_model_name = 'Random Forest Classifier'
best_classifier = classification_models[best_model_name]

# Save the best model to a .pkl file
model_filename = f'{best_model_name.replace(" ", "_").lower()}_classification_model.pkl'
joblib.dump(best_classifier, model_filename)

print(f"Best classification model ('{best_model_name}') saved to '{model_filename}'")
```

## Hyperparameter Tuning for Random Forest Classifier

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report # Ensure these are imported if not already

# Define a smaller parameter grid for Random Forest
param_grid = {
    'n_estimators': [100, 150],
    'max_features': ['sqrt'],
    'max_depth': [10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'bootstrap': [True]
}

# Initialize a RandomForestClassifier
rf_clf = RandomForestClassifier(random_state=42)

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=rf_clf, param_grid=param_grid,
                           cv=3, n_jobs=-1, verbose=2, scoring='accuracy')

# Fit GridSearchCV to the resampled training data (no leaking features)
print("Starting GridSearchCV for Random Forest Classifier with reduced parameter grid...")
grid_search.fit(x_train_clf_resampled_no_leak, y_train_clf_resampled_no_leak)
print("GridSearchCV completed.")
```

### Resuming Hyperparameter Tuning

```python
# The GridSearchCV was previously interrupted. Re-running the fit method to complete the tuning.
print("Resuming GridSearchCV for Random Forest Classifier...")
grid_search.fit(x_train_clf_resampled_no_leak, y_train_clf_resampled_no_leak)
print("GridSearchCV completed.")
```

```python
# Print the best parameters and best score
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best accuracy score: {grid_search.best_score_:.2f}")

# Get the best estimator
best_rf_model = grid_search.best_estimator_

# Evaluate the best model on the test set
y_pred_best_rf = best_rf_model.predict(x_test_clf_no_leak)
accuracy_best_rf = accuracy_score(y_test_clf_no_leak, y_pred_best_rf)

print(f"\nAccuracy of the best Random Forest model on the test set: {accuracy_best_rf:.2f}")
print(f"Classification Report for the best Random Forest model:\n{classification_report(y_test_clf_no_leak, y_pred_best_rf, target_names=le.classes_)}")
```
